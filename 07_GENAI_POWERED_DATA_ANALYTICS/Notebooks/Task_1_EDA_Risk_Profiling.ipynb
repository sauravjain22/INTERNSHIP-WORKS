## 1.3 Missing Data Proportion

The number of missing values alone does not show how significant a data-quality issue is. Calculating the percentage of missing values relative to the total number of records helps assess the extent of missingness in each affected variable.

```python
# Calculate the percentage of missing values in each column

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_percentage[missing_percentage > 0]
```

## 1.4 Summary Statistics of Key Financial Variables

Before deciding how to handle missing values, the distributions of the affected numerical variables should be examined. Summary statistics such as the mean, median, minimum, maximum, and quartiles can help identify skewness and unusual values that may influence the choice of imputation method.

```python
# Generate summary statistics for variables containing missing values

df[["Income", "Credit_Score", "Loan_Balance"]].describe()
```

### Interpretation

The summary statistics show that all three variables have different distributions and scales.

`Income` has a mean of approximately 108,380 and a median of 107,658. The mean and median are relatively close, suggesting that the distribution is not strongly skewed based on these summary statistics.

`Credit_Score` has a mean of approximately 577.7 and a median of 586. The values range from 301 to 847, and the mean and median are also relatively close.

`Loan_Balance` has a mean of approximately 48,654 and a median of 45,776. The mean is somewhat higher than the median, which may indicate some right-skewness in the distribution.

These results suggest that the missing values should not automatically be replaced using the mean. The median may provide a more robust imputation method, particularly for variables where extreme values or skewness may affect the mean.

## 2. Missing Data Treatment

Based on the missing-data proportions and summary statistics, median imputation will be used for `Income`, `Credit_Score`, and `Loan_Balance`.

Median imputation preserves all customer records and is less sensitive to extreme values than mean imputation. Synthetic data generation is not required at this stage because the amount of missing data is limited and can be addressed using a transparent statistical method.

```python
# Fill missing values using the median of each respective variable

df["Income"] = df["Income"].fillna(df["Income"].median())
df["Credit_Score"] = df["Credit_Score"].fillna(df["Credit_Score"].median())
df["Loan_Balance"] = df["Loan_Balance"].fillna(df["Loan_Balance"].median())
```

```python
# Verify that no missing values remain in the treated variables

df[["Income", "Credit_Score", "Loan_Balance"]].isnull().sum()
```

### Verification

After applying median imputation, `Income`, `Credit_Score`, and `Loan_Balance` contain no remaining missing values. This confirms that the selected missing-data treatment was successfully applied and the dataset is ready for the next stage of data-quality assessment.

## 3.1 Missed Payments and Delinquency

Missed payment history is an important potential risk indicator because repeated missed payments may be associated with delinquent accounts.

The relationship between `Missed_Payments` and `Delinquent_Account` will be examined to determine whether delinquency rates differ across customers with different numbers of missed payments.

```python
# Calculate the number of customers and delinquency rate for each missed-payment level

missed_payment_risk = df.groupby("Missed_Payments")["Delinquent_Account"].agg(
    Customer_Count="count",
    Delinquency_Rate="mean"
)

missed_payment_risk["Delinquency_Rate"] *= 100

missed_payment_risk
```

### Interpretation

The observed delinquency rate varies across different levels of missed payments rather than increasing consistently with the number of missed payments.

Customers with 5 missed payments have the highest observed delinquency rate at approximately 22.06%, while customers with 6 missed payments have a lower rate of approximately 11.48%. Customers with 1 missed payment have a delinquency rate of approximately 18.52%, compared with 18.18% for customers with no missed payments.

This indicates that the relationship between `Missed_Payments` and `Delinquent_Account` is not strictly monotonic in this dataset. The unexpected decrease at 6 missed payments should be investigated further, as it may reflect the relatively small group sizes, characteristics of the underlying customers, or other variables affecting delinquency. Therefore, `Missed_Payments` should be considered a potentially relevant risk indicator, but the observed results do not establish a simple linear relationship with delinquency.

## 3.2 Credit Utilization and Delinquency

Credit utilization represents the proportion of available credit being used by a customer. It is considered a potential risk indicator because higher utilization may be associated with greater financial pressure.

The relationship between `Credit_Utilization` and `Delinquent_Account` will be examined to determine whether customers with different utilization levels show different observed delinquency rates.

```python
# Examine the distribution of credit utilization

df["Credit_Utilization"].describe()
```

### Interpretation

The `Credit_Utilization` variable has 500 observations, with a mean of approximately 0.491 and a median of approximately 0.486. This indicates that the typical customer is using around half of their available credit.

The values range from approximately 0.05 to 1.026. Most customers fall between the 25th percentile of 0.356 and the 75th percentile of 0.634. The maximum value slightly exceeds 1.0, indicating that at least one customer has credit utilization above 100%.

The mean and median are relatively close, suggesting that the overall distribution is not strongly skewed. However, the value above 1.0 should be investigated because very high credit utilization may represent customers with unusually high credit usage or a potential data-quality issue.

Credit utilization will therefore be examined further by comparing utilization levels with `Delinquent_Account` to determine whether higher utilization is associated with different observed delinquency rates.

```python
# Compare average credit utilization between delinquent and non-delinquent customers
credit_utilization_risk = df.groupby("Delinquent_Account")["Credit_Utilization"].agg(
    Customer_Count="count",
    Average_Utilization="mean",
    Median_Utilization="median"
)

credit_utilization_risk
```

### Interpretation

The comparison shows that delinquent customers have a slightly higher average credit utilization than non-delinquent customers.

- Non-delinquent customers: average utilization ≈ 0.489
- Delinquent customers: average utilization ≈ 0.507

The median utilization follows the same pattern:

- Non-delinquent customers: median ≈ 0.484
- Delinquent customers: median ≈ 0.493

This indicates a small positive association between credit utilization and observed delinquency in this dataset. However, the difference is relatively small, so credit utilization alone does not appear to strongly distinguish delinquent customers from non-delinquent customers.

Therefore, `Credit_Utilization` should be considered as one potential risk factor alongside other financial and behavioural variables rather than as a standalone indicator of delinquency.

## 3.3 Credit Score and Delinquency

## 3.6 Monthly Payment Behaviour and Delinquency

The dataset contains six monthly payment-status variables representing customer payment behaviour over time.

Each month contains three possible statuses: `On-time`, `Late`, and `Missed`. Examining these variables against `Delinquent_Account` can help identify whether particular payment behaviours are associated with higher observed delinquency rates.

The analysis will compare delinquency rates across the payment-status categories for each month.

```python
# Calculate delinquency rates for each monthly payment status
monthly_payment_risk = {}

for month in [f"Month_{i}" for i in range(1, 7)]:
    monthly_payment_risk[month] = (
        df.groupby(month)["Delinquent_Account"]
        .agg(Customer_Count="count", Delinquency_Rate="mean")
    )

    monthly_payment_risk[month]["Delinquency_Rate"] *= 100

monthly_payment_risk
```

### Interpretation

The monthly payment analysis shows that delinquency rates vary across payment-status categories and months, but there is no consistent pattern in which `Missed` payments always correspond to the highest delinquency rate.

For example, in Month 1, customers with `On-time` payments have the highest observed delinquency rate (19.77%), compared with 16.46% for `Missed` and 11.32% for `Late`. In Month 4, the pattern changes, with `Late` payments having the highest delinquency rate (19.34%), while `On-time` payments have the lowest (11.32%).

Across the six months, the highest observed rate for a payment-status group is 19.77% for `On-time` customers in Month 1, while the lowest is 11.32% for `Late` customers in Month 1 and `On-time` customers in Month 4. These unexpected variations suggest that individual monthly payment status alone does not provide a simple or consistent explanation of delinquency in this dataset.

Monthly payment behaviour should therefore be retained as a potential predictive feature, but the observed patterns should be interpreted cautiously and investigated together with other variables. The unexpected relationship between payment status and delinquency is an important finding for the analytics team to investigate further.

## 3.7 Overall Missed-Payment Behaviour

The six monthly payment-status variables can be combined to identify the number of months in which each customer had a `Missed` payment.

This provides a simple measure of repeated missed-payment behaviour and allows us to examine whether customers with more frequent missed payments have different observed delinquency rates.

```python
# Count the number of months with a missed payment for each customer
month_columns = [f"Month_{i}" for i in range(1, 7)]

df["Missed_Months"] = (df[month_columns] == "Missed").sum(axis=1)

# Compare delinquency rates across the number of missed-payment months
missed_months_risk = df.groupby("Missed_Months")["Delinquent_Account"].agg(
    Customer_Count="count",
    Delinquency_Rate="mean"
)

missed_months_risk["Delinquency_Rate"] *= 100

missed_months_risk
```

### Interpretation

The number of months with missed payments shows variation in observed delinquency rates, but the relationship is not consistently increasing as the number of missed months increases.

Customers with 0 missed months have an observed delinquency rate of 20.00%, while customers with 1, 2, 3, and 4 missed months have rates of approximately 13.71%, 17.51%, 13.79%, and 17.65%, respectively. The group with 5 missed months has the highest observed delinquency rate at 25.00%; however, this group contains only 8 customers, so the result should be interpreted cautiously. The 6-month group contains only one customer and has a 0% observed delinquency rate, making it unsuitable for drawing conclusions.

Overall, repeated missed-payment behaviour appears potentially relevant to delinquency prediction, but the relationship is not monotonic in this dataset. The small group sizes at the highest missed-month levels are an important limitation and should be considered when developing and evaluating a predictive model.

### Overall EDA Conclusion

The analysis indicates that delinquency is not explained by a single variable. Credit utilization and DTI show small differences between delinquent and non-delinquent customers, while payment behaviour and missed-payment measures show inconsistent relationships.
