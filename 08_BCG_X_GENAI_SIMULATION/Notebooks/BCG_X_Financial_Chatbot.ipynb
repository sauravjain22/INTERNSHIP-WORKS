## 1. Chatbot Prototype

This notebook develops a simplified rule-based financial chatbot using the financial dataset analyzed in Task 1.

The prototype focuses on three core chatbot principles:

- Financial data retrieval

## 2. Environment and Data Preparation

The chatbot uses pandas to load and access the financial dataset.

The dataset contains financial information for Microsoft, Tesla, and Apple for fiscal years 2021–2025.

```python
import pandas as pd

# Load the financial dataset
df = pd.read_csv(
    "../data/raw/BCG_X_Financial_Dataset_5_Years_DEMO.csv"
)

# Display the first five records
df.head()
```

## 3. Financial Data Retrieval

The chatbot needs a simple retrieval mechanism to access the correct financial value from the dataset.

The retrieval function accepts a company, fiscal year, and financial metric, then filters the DataFrame to find the corresponding value.

This separates data retrieval from chatbot logic and makes the chatbot easier to maintain and extend.

```python
def get_financial_value(company, year, metric):
    """
    Retrieve a financial metric for a specific company and fiscal year.
    """
    
    result = df[
        (df["Company"].str.lower() == company.lower()) &
        (df["Fiscal Year"] == year)
    ]

    if result.empty:
        return None

    return result.iloc[0][metric]
```

```python
# Test the retrieval function
revenue = get_financial_value(
    "Apple",
    2025,
    "Total Revenue (USD millions)"
)

print("Apple's 2025 revenue:", revenue, "USD million")
```

## 4. Rule-Based Chatbot Logic

The chatbot uses if-else logic to match a user's input with predefined financial queries.

When a supported query is recognized, the chatbot retrieves the required financial value from the dataset and generates a response.

If the query is not recognized, the chatbot will provide a helpful message explaining that only predefined financial queries are currently supported.

```python
def simple_chatbot(user_query):
    """
    Respond to predefined financial queries.
    """
    
    user_query = user_query.strip().lower()

    # Query 1: Apple's revenue in 2025
    if user_query == "what was apple's revenue in 2025?":
        
        revenue = get_financial_value(
            "Apple",
            2025,
            "Total Revenue (USD millions)"
        )
        
        return f"Apple's revenue in 2025 was USD {revenue:,.0f} million."

    # Query 2: Microsoft's net income in 2025
    elif user_query == "what was microsoft's net income in 2025?":
        
        net_income = get_financial_value(
            "Microsoft",
            2025,
            "Net Income (USD millions)"
        )
        
        return f"Microsoft's net income in 2025 was USD {net_income:,.0f} million."

    # Query 3: Tesla's operating cash flow in 2025
    elif user_query == "what was tesla's operating cash flow in 2025?":
        
        cash_flow = get_financial_value(
            "Tesla",
            2025,
            "Operating Cash Flow (USD millions)"
        )
        
        return f"Tesla's operating cash flow in 2025 was USD {cash_flow:,.0f} million."

    # Query 4: Company with highest revenue in 2025
    elif user_query == "which company had the highest revenue in 2025?":
        
        revenue_2025 = df[df["Fiscal Year"] == 2025]
        
        highest_company = revenue_2025.loc[
            revenue_2025["Total Revenue (USD millions)"].idxmax(),
            "Company"
        ]
        
        highest_revenue = revenue_2025[
            "Total Revenue (USD millions)"
        ].max()
        
        return (
            f"{highest_company} had the highest revenue in 2025, "
            f"at USD {highest_revenue:,.0f} million."
        )

    # Query 5: Tesla's net income year-over-year change
    elif user_query == "how has tesla's net income changed over the last year?":
        
        net_income_2024 = get_financial_value(
            "Tesla",
            2024,
            "Net Income (USD millions)"
        )
        
        net_income_2025 = get_financial_value(
            "Tesla",
            2025,
            "Net Income (USD millions)"
        )
        
        change = net_income_2025 - net_income_2024
        percentage_change = (change / net_income_2024) * 100
        
        direction = "increased" if change > 0 else "decreased"
        
        return (
            f"Tesla's net income {direction} from "
            f"USD {net_income_2024:,.0f} million in 2024 to "
            f"USD {net_income_2025:,.0f} million in 2025. "
            f"This represents a change of {percentage_change:.2f}%."
        )

    # Unsupported query
    else:
        return (
            "Sorry, I can only answer predefined financial queries "
            "about Microsoft, Tesla, and Apple."
        )
```

```python
response = simple_chatbot(
    "What was Apple's revenue in 2025?"
)

print(response)
```

```python
response = simple_chatbot(
    "Tell me about Apple's stock price."
)

print(response)
```

```python
test_queries = [
    "What was Apple's revenue in 2025?",
    "What was Microsoft's net income in 2025?",
    "What was Tesla's operating cash flow in 2025?",
    "Which company had the highest revenue in 2025?",
    "How has Tesla's net income changed over the last year?"
]

for query in test_queries:
    print("User:", query)
    print("Bot:", simple_chatbot(query))
    print()
```

## 5. Interactive Chatbot

The chatbot can now be used interactively through the Python `input()` function.

Users can enter one of the predefined financial queries, and the chatbot will process the query and return the corresponding financial response.

The chatbot also provides an exit option so the user can end the conversation.

```python
print("Financial Chatbot")
print("Type 'exit' to end the conversation.\n")

while True:
    user_query = input("You: ")

    if user_query.strip().lower() == "exit":
        print("Bot: Thank you for using the financial chatbot.")
        break

    response = simple_chatbot(user_query)
    print("Bot:", response)
```

## 7. Chatbot Functionality

The financial chatbot is a rule-based prototype that responds to five predefined financial queries.

### Supported Queries

1. What was Apple's revenue in 2025?
2. What was Microsoft's net income in 2025?
3. What was Tesla's operating cash flow in 2025?
4. Which company had the highest revenue in 2025?
5. How has Tesla's net income changed over the last year?

### How It Works

The chatbot receives a user's query and compares it with predefined query patterns using if-else logic.

When a matching query is identified, the chatbot retrieves the relevant financial data from the pandas DataFrame and generates a response.

For comparison and year-over-year queries, the chatbot performs calculations using the financial dataset before generating the response.

Unsupported queries return a predefined error message explaining that the chatbot currently supports only the predefined financial questions.
