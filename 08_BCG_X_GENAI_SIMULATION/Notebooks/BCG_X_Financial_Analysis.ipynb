## 6. Data Sorting and Preparation for Financial Analysis

Before calculating year-over-year financial growth, the dataset must be arranged in chronological order for each company.

Sorting ensures that each fiscal year's financial figures are compared with the preceding fiscal year's figures for the same company. This is essential for accurate growth calculations using pandas.

The dataset will be sorted by company name and fiscal year, and the index will be reset to maintain a clean sequential index.

```python
# Sort records by company and fiscal year
df = df.sort_values(
    by=["Company", "Fiscal Year"]
).reset_index(drop=True)

# Display the sorted dataset
df
```

## 7. Year-over-Year (YoY) Growth Analysis

Year-over-year growth measures the percentage change in a financial metric compared with the previous fiscal year.

It helps analysts understand whether a company's financial performance is improving, declining, or remaining relatively stable over time.

### Revenue Growth

Revenue growth measures the percentage change in total revenue compared with the previous fiscal year.

**Formula:**

Revenue Growth (%) = ((Current Year Revenue − Previous Year Revenue) / Previous Year Revenue) × 100

### Net Income Growth

Net income growth measures the percentage change in net income compared with the previous fiscal year.

**Formula:**

Net Income Growth (%) = ((Current Year Net Income − Previous Year Net Income) / Previous Year Net Income) × 100

### Interpretation

- **Positive growth:** The financial metric increased compared with the previous year.
- **Negative growth:** The financial metric decreased compared with the previous year.
- **Zero growth:** The financial metric remained unchanged.
- **NaN for the first year:** Growth cannot be calculated because the previous year's data is unavailable.

Growth will be calculated separately for each company to ensure that one company's financial results are never compared with another company's results.

```python
# Calculate year-over-year revenue growth
df["Revenue Growth (%)"] = (
    df.groupby("Company")["Total Revenue (USD millions)"]
    .pct_change() * 100
)

# Calculate year-over-year net income growth
df["Net Income Growth (%)"] = (
    df.groupby("Company")["Net Income (USD millions)"]
    .pct_change() * 100
)

# Display the results
df[
    [
        "Company",
        "Fiscal Year",
        "Total Revenue (USD millions)",
        "Revenue Growth (%)",
        "Net Income (USD millions)",
        "Net Income Growth (%)"
    ]
]
```

## 8. Comparative Financial Performance Analysis

This section compares the financial performance of Microsoft, Tesla, and Apple over the five-year period from 2021 to 2025.

The analysis focuses on revenue growth, net income growth, and changes in profitability trends.

The comparison will help identify differences in growth patterns, highlight significant changes in financial performance, and establish a foundation for further financial analysis.

The findings are based on the demonstration dataset and should be interpreted as educational observations rather than verified financial conclusions.

```python
# Calculate average annual growth rates for each company
growth_summary = (
    df.groupby("Company")[
        ["Revenue Growth (%)", "Net Income Growth (%)"]
    ]
    .mean()
    .round(2)
)

# Display the comparative growth summary
growth_summary
```

## 9. Financial Metrics Comparison

Growth rates show how quickly financial metrics change, but they do not show the overall scale of a company's operations or financial position.

This section compares five key financial metrics across Microsoft, Tesla, and Apple:

- Total Revenue
- Net Income
- Total Assets
- Total Liabilities
- Operating Cash Flow

Average values across the available fiscal years will be calculated for each company. This comparison provides a broader view of operating performance, profitability, asset size, financial obligations, and cash generation.

All values are expressed in USD millions and are based on the demonstration dataset.

```python
# Calculate average financial metrics for each company
financial_summary = (
    df.groupby("Company")[
        [
            "Total Revenue (USD millions)",
            "Net Income (USD millions)",
            "Total Assets (USD millions)",
            "Total Liabilities (USD millions)",
            "Operating Cash Flow (USD millions)"
        ]
    ]
    .mean()
    .round(2)
)

# Display the financial comparison
financial_summary
```

## 10. Financial Data Visualization

Visualizations provide a clear way to compare financial performance across companies and identify patterns that may not be immediately apparent in tables.

This section uses bar charts to compare the average revenue, net income, and operating cash flow of Microsoft, Tesla, and Apple.

These visualizations support the numerical analysis and help communicate the findings more effectively.

```python
import matplotlib.pyplot as plt

# Plot average revenue by company
financial_summary["Total Revenue (USD millions)"].plot(
    kind="bar",
    figsize=(8, 5),
    title="Average Annual Revenue by Company"
)

plt.xlabel("Company")
plt.ylabel("Revenue (USD millions)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Create two side-by-side financial comparison charts
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Net income comparison
financial_summary["Net Income (USD millions)"].plot(
    kind="bar",
    ax=axes[0],
    title="Average Annual Net Income by Company"
)

axes[0].set_xlabel("Company")
axes[0].set_ylabel("Net Income (USD millions)")
axes[0].tick_params(axis="x", rotation=0)

# Operating cash flow comparison
financial_summary["Operating Cash Flow (USD millions)"].plot(
    kind="bar",
    ax=axes[1],
    title="Average Annual Operating Cash Flow by Company"
)

axes[1].set_xlabel("Company")
axes[1].set_ylabel("Operating Cash Flow (USD millions)")
axes[1].tick_params(axis="x", rotation=0)

plt.tight_layout()
plt.show()
```
